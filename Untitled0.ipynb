#1. Implementação dos métodos numéricos

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import display, HTML


def bissecao(f, a, b, tol=1e-6, max_iter=100):
    """
    Método da Bisseção com verificação de erro e critério de parada duplo.
    """
    if f(a) * f(b) > 0:
        return None, 0, [], "Erro: f(a) e f(b) possuem o mesmo sinal."

    historico = []
    c_ant = a

    for i in range(1, max_iter + 1):
        c = (a + b) / 2.0
        historico.append((a, b, c))

        # Critério de parada: tolerância de x e resíduo |f(x)|
        if i > 1 and abs(c - c_ant) < tol and abs(f(c)) < tol:
            return c, i, historico, "Convergiu"
        if abs(f(c)) < tol:
            return c, i, historico, "Convergiu"

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

        c_ant = c

    return c, max_iter, historico, "Falha: Atingiu o limite de iterações."


def newton(f, df, x0, tol=1e-6, max_iter=100):
    """
    Método de Newton-Raphson com proteção contra derivada nula.
    """
    historico = [x0]
    xk = x0

    for i in range(1, max_iter + 1):
        df_x = df(xk)

        # Proteção contra derivada zero
        if abs(df_x) < 1e-12:
            return xk, i, historico, "Falha: Derivada nula ou próxima de zero."

        x_next = xk - f(xk) / df_x
        historico.append(x_next)

        if abs(x_next - xk) < tol and abs(f(x_next)) < tol:
            return x_next, i, historico, "Convergiu"

        xk = x_next

    return xk, max_iter, historico, "Falha: Atingiu o limite de iterações."


def secante(f, x0, x1, tol=1e-6, max_iter=100):
    """
    Método da Secante com proteção contra divisão por zero.
    """
    historico = [x0, x1]

    for i in range(1, max_iter + 1):
        f_x0, f_x1 = f(x0), f(x1)
        denom = f_x1 - f_x0

        # Proteção contra divisão por zero
        if abs(denom) < 1e-12:
            return x1, i, historico, "Falha: Denominador próximo de zero."

        x2 = x1 - f_x1 * (x1 - x0) / denom
        historico.append(x2)

        if abs(x2 - x1) < tol and abs(f(x2)) < tol:
            return x2, i, historico, "Convergiu"

        x0, x1 = x1, x2

    return x1, max_iter, historico, "Falha: Atingiu o limite de iterações."
```

#2. Definição das funções

```python
funcoes = {
    "f1": {
        "f": lambda x: x**2 - 2,
        "df": lambda x: 2*x,
        "bissecao": [1.0, 2.0],
        "newton": 1.5,
        "secante": (1.0, 2.0),
        "nome": "f₁(x) = x² - 2"
    },
    "f2": {
        "f": lambda x: x**3 - x - 2,
        "df": lambda x: 3*x**2 - 1,
        "bissecao": [1.0, 2.0],
        "newton": 1.5,
        "secante": (1.0, 2.0),
        "nome": "f₂(x) = x³ - x - 2"
    },
    "f3": {
        "f": lambda x: np.exp(-x) - x,
        "df": lambda x: -np.exp(-x) - 1,
        "bissecao": [0.0, 1.0],
        "newton": 0.5,
        "secante": (0.0, 1.0),
        "nome": "f₃(x) = e⁻ˣ - x"
    },
    "f4": {
        "f": lambda x: x**3 - 2*x + 2,
        "df": lambda x: 3*x**2 - 2,
        "bissecao": [-2.0, -1.0],
        "newton": -2.0,
        "secante": (-2.0, -1.0),
        "nome": "f₄(x) = x³ - 2x + 2"
    }
}
```

#3. Gerações e Exibição da tabela formatada

```python
resultados = []

for chave, dados in funcoes.items():
    f, df = dados["f"], dados["df"]

    # Execução - Bisseção
    a, b = dados["bissecao"]
    r, iters, hist, status = bissecao(f, a, b)
    resultados.append({
        "Função": dados["nome"],
        "Método": "Bisseção",
        "Valores Iniciais": f"[{a}, {b}]",
        "Raiz Aproximada": f"{r:.7f}" if r is not None else "N/A",
        "|f(x)| Final": f"{abs(f(r)):.2e}" if r is not None else "N/A",
        "Resultado": f"{iters} iterações ({status})"
    })

    # Execução - Newton
    x0 = dados["newton"]
    r, iters, hist, status = newton(f, df, x0)
    resultados.append({
        "Função": dados["nome"],
        "Método": "Newton",
        "Valores Iniciais": f"x₀ = {x0}",
        "Raiz Aproximada": f"{r:.7f}" if r is not None else "N/A",
        "|f(x)| Final": f"{abs(f(r)):.2e}" if r is not None else "N/A",
        "Resultado": f"{iters} iterações ({status})"
    })

    # Execução - Secante
    x0, x1 = dados["secante"]
    r, iters, hist, status = secante(f, x0, x1)
    resultados.append({
        "Função": dados["nome"],
        "Método": "Secante",
        "Valores Iniciais": f"({x0}, {x1})",
        "Raiz Aproximada": f"{r:.7f}" if r is not None else "N/A",
        "|f(x)| Final": f"{abs(f(r)):.2e}" if r is not None else "N/A",
        "Resultado": f"{iters} iterações ({status})"
    })

# Criar DataFrame com o Pandas
df_resultados = pd.DataFrame(resultados)

# Exibir como uma TABELA visual e estilizada no Google Colab
print("--- TABELA DE RESULTADOS ---")
display(df_resultados.style.set_properties(**{
    'background-color': '#f8f9fa',
    'color': '#212529',
    'border-color': '#dee2e6',
    'text-align': 'center'
}).set_table_styles([
    {'selector': 'th', 'props': [('background-color', '#0d6efd'), ('color', 'white'), ('font-weight', 'bold'), ('text-align', 'center')]}
]))
```

# 4. Geração e salvamento dos GIFS

```python

print("\nGerando animações em GIF...")

def criar_gif_bissecao(chave="f1", filename="bissecao_f1.gif"):
    dados = funcoes[chave]
    f = dados["f"]
    a_init, b_init = dados["bissecao"]
    _, _, historico, _ = bissecao(f, a_init, b_init)

    fig, ax = plt.subplots(figsize=(7, 5))
    x_vals = np.linspace(a_init - 0.5, b_init + 0.5, 400)

    def update(k):
        ax.clear()
        a, b, c = historico[k]
        ax.plot(x_vals, f(x_vals), label=f'{dados["nome"]}', color='blue')
        ax.axhline(0, color='black', linestyle='--', linewidth=0.8)
        ax.axvspan(a, b, color='yellow', alpha=0.3, label='Intervalo [a, b]')
        ax.plot(c, f(c), 'ro', label=f'x_k = {c:.5f}')
        ax.set_title(f'Bisseção - Iteração {k+1}\n|f(x_k)| = {abs(f(c)):.2e}')
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.legend(loc='upper left')
        ax.grid(True)

    anim = animation.FuncAnimation(fig, update, frames=len(historico), interval=800)
    anim.save(filename, writer='pillow')
    plt.close()


def criar_gif_newton(chave="f1", filename="newton_f1.gif"):
    dados = funcoes[chave]
    f, df = dados["f"], dados["df"]
    x0 = dados["newton"]
    _, _, historico, _ = newton(f, df, x0)

    fig, ax = plt.subplots(figsize=(7, 5))
    x_vals = np.linspace(min(historico) - 0.5, max(historico) + 0.5, 400)

    def update(k):
        ax.clear()
        xk = historico[k]
        ax.plot(x_vals, f(x_vals), label=f'{dados["nome"]}', color='blue')
        ax.axhline(0, color='black', linestyle='--', linewidth=0.8)

        x_tangent = np.linspace(xk - 0.5, xk + 0.5, 100)
        y_tangent = f(xk) + df(xk) * (x_tangent - xk)
        ax.plot(x_tangent, y_tangent, 'r--', label='Reta Tangente')
        ax.plot(xk, f(xk), 'ro', label=f'x_{k} = {xk:.5f}')

        ax.set_title(f'Newton-Raphson - Iteração {k}\n|f(x_k)| = {abs(f(xk)):.2e}')
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.legend(loc='upper left')
        ax.grid(True)

    anim = animation.FuncAnimation(fig, update, frames=len(historico), interval=800)
    anim.save(filename, writer='pillow')
    plt.close()


def criar_gif_secante(chave="f1", filename="secante_f1.gif"):
    dados = funcoes[chave]
    f = dados["f"]
    x0, x1 = dados["secante"]
    _, _, historico, _ = secante(f, x0, x1)

    fig, ax = plt.subplots(figsize=(7, 5))
    x_vals = np.linspace(min(historico) - 0.5, max(historico) + 0.5, 400)

    def update(k):
        if k < 1:
            return
        ax.clear()
        xk_prev, xk = historico[k-1], historico[k]
        ax.plot(x_vals, f(x_vals), label=f'{dados["nome"]}', color='blue')
        ax.axhline(0, color='black', linestyle='--', linewidth=0.8)

        slope = (f(xk) - f(xk_prev)) / (xk - xk_prev)
        x_sec = np.linspace(min(xk_prev, xk) - 0.3, max(xk_prev, xk) + 0.3, 100)
        y_sec = f(xk) + slope * (x_sec - xk)

        ax.plot(x_sec, y_sec, 'g--', label='Reta Secante')
        ax.plot([xk_prev, xk], [f(xk_prev), f(xk)], 'go')

        ax.set_title(f'Secante - Passo {k}\n|f(x_k)| = {abs(f(xk)):.2e}')
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.legend(loc='upper left')
        ax.grid(True)

    anim = animation.FuncAnimation(fig, update, frames=len(historico), interval=800)
    anim.save(filename, writer='pillow')
    plt.close()


# Execução da geração dos 3 arquivos de GIF
criar_gif_bissecao("f1", "bissecao_f1.gif")
criar_gif_newton("f1", "newton_f1.gif")
criar_gif_secante("f1", "secante_f1.gif")

print("✅ Todos os GIFs foram gerados e salvos com sucesso na pasta de arquivos do Colab!")
```
